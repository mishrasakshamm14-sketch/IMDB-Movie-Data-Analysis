# Additional IMDB Movie Data Analysis
## Analysis performed by me

TOP TEN COMEDY MOVIES

```python
top_comedy = data[
    data['Genre'].str.contains(
        'Comedy', case=False, na=False
    )
].sort_values(
    by='Rating', ascending=False
).head(10)

print("Top 10 Highest Rated Comedy Movies")

display(top_comedy[['Title', 'Genre', 'Rating']])
```

Top 5 Highest-Rated Movies Overall

```python
top_5 = data.sort_values(
    by='Rating',
    ascending=False
).head(5)

print("Top 5 Highest Rated Movies Overall")

display(top_5[['Title', 'Genre', 'Rating']])
```

GRAPH FOR COMEDY MOVIES ANALYSIS

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.barh(
    top_comedy['Title'],
    top_comedy['Rating'],
    color='skyblue'
)

plt.xlabel('IMDB Rating')
plt.ylabel('Movie Title')
plt.title('Top 10 Highest Rated Comedy Movies')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

GRAPH FOR MOVIE GENERS WHICH ARE MOST COMMON

```python
genre_counts = (
    data['Genre']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12, 6))

genre_counts.plot(
    kind='bar',
    color='cornflowerblue'
)

plt.title('Top 10 Most Common Movie Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Top 10 highest-rated superhero movies

```python
import matplotlib.pyplot as plt

superhero_keywords = (
    r'Avengers|Batman|Spider-Man|Spiderman|'
    r'Superman|Iron Man|Captain America|Thor|'
    r'Wonder Woman|Deadpool|X-Men|Wolverine|'
    r'Guardians of the Galaxy|Doctor Strange|'
    r'Ant-Man|Black Panther|Suicide Squad|'
    r'Justice League|Logan'
)

superhero = data[
    data['Title'].str.contains(
        superhero_keywords,
        case=False,
        na=False
    )
].sort_values(
    'Rating', ascending=False
).head(10)

print("Top 10 Highest Rated Superhero Movies")

display(superhero[['Title', 'Year', 'Rating']])

plt.figure(figsize=(12, 6))

plt.barh(
    superhero['Title'],
    superhero['Rating'],
    color='purple'
)

plt.title('Top 10 Highest Rated Superhero Movies')
plt.xlabel('IMDB Rating')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

Do higher-budget movies earn more revenue?

```python
rating_revenue = data[
    ['Rating', 'Revenue (Millions)']
].dropna()

plt.figure(figsize=(10, 6))

plt.scatter(
    rating_revenue['Rating'],
    rating_revenue['Revenue (Millions)'],
    alpha=0.5,
    color='darkorange'
)

plt.title('IMDB Ratings vs Movie Revenue')
plt.xlabel('IMDB Rating')
plt.ylabel('Revenue (Millions)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

correlation = rating_revenue[
    'Rating'
].corr(
    rating_revenue['Revenue (Millions)']
)

print(
    "Correlation between Rating and Revenue:",
    round(correlation, 3)
)
```
